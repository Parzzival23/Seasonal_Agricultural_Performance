```python
# ============================================================
# SEASONAL AGRICULTURE PERFORMANCE ANALYSIS
# VOIS AICTE BATCH 1 2026-2027
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from itertools import combinations
from google.colab import files

warnings.filterwarnings("ignore")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

# Visualization settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("\nUpload your agricultural CSV dataset.")

uploaded = files.upload()

# Find uploaded CSV
csv_files = [file for file in os.listdir("/content") if file.endswith(".csv")]

if len(csv_files) == 0:
    raise FileNotFoundError("No CSV file found.")

file_path = "/content/" + csv_files[0]

# Load dataset
df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")
print("File:", csv_files[0])
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# 3. PROJECT INTRODUCTION
# ============================================================

print("\n" + "=" * 70)
print("SEASONAL AGRICULTURE PERFORMANCE ANALYSIS")
print("=" * 70)

print("""
Objective:
Analyze agricultural performance across different seasons and identify
meaningful patterns, trends, relationships, and variations.

The analysis focuses on:
- Agricultural yield and production
- Environmental conditions
- Soil characteristics
- Farming practices
- Resource usage
- Water efficiency
- Economic performance
- Disease and pest risk
- Regional and crop-level differences
""")


# ============================================================
# 4. INITIAL DATA EXPLORATION
# ============================================================

print("\n" + "=" * 70)
print("INITIAL DATA EXPLORATION")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())


# ============================================================
# 5. COLUMN INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("COLUMN INFORMATION")
print("=" * 70)

column_info = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique()
})

display(column_info)


# ============================================================
# 6. CATEGORICAL VARIABLES
# ============================================================

print("\n" + "=" * 70)
print("CATEGORICAL VARIABLES")
print("=" * 70)

categorical_columns = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(list(categorical_columns))

for col in categorical_columns:
    print("\n" + "-" * 50)
    print(col)
    print("-" * 50)
    print(df[col].value_counts())


# ============================================================
# 7. NUMERICAL VARIABLES
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL VARIABLES")
print("=" * 70)

numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(list(numerical_columns))


# ============================================================
# 8. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing = df.isnull().sum().sort_values(ascending=False)

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": (missing / len(df) * 100).round(2)
})

display(missing_df[missing_df["Missing Values"] > 0])

# Missing value visualization
missing_nonzero = missing[missing > 0]

if len(missing_nonzero) > 0:
    plt.figure(figsize=(10, 5))

    plt.bar(
        missing_nonzero.index,
        missing_nonzero.values
    )

    plt.title("Missing Values by Column")
    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")


# ============================================================
# 9. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")


# ============================================================
# 10. DATA CLEANING
# ============================================================

print("\n" + "=" * 70)
print("DATA CLEANING")
print("=" * 70)

# Columns with missing numeric values
numeric_imputation_columns = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for col in numeric_imputation_columns:

    if col in df.columns:
        missing_before = df[col].isnull().sum()

        if missing_before > 0:
            median_value = df[col].median()
            df[col] = df[col].fillna(median_value)

            print(
                f"{col}: Filled {missing_before} missing values "
                f"using median = {median_value:.2f}"
            )

print("\nTotal missing values after cleaning:")
print(df.isnull().sum().sum())


# ============================================================
# 11. VERIFY CLEAN DATASET
# ============================================================

print("\n" + "=" * 70)
print("CLEAN DATASET VERIFICATION")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())


# ============================================================
# 12. UNIQUE CATEGORY SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MAIN CATEGORIES")
print("=" * 70)

important_categorical = [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]

for col in important_categorical:

    if col in df.columns:
        print(f"\n{col}:")
        print(df[col].unique())


# ============================================================
# 13. FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)

# Profit margin
df["Profit_Margin_pct"] = (
    df["Profit_INR"] / df["Revenue_INR"] * 100
)

# Revenue per hectare
df["Revenue_per_Hectare_INR"] = (
    df["Revenue_INR"] / df["Farm_Area_Hectares"]
)

# Cost per hectare
df["Cost_per_Hectare_INR"] = (
    df["Total_Cost_INR"] / df["Farm_Area_Hectares"]
)

# Profit per hectare
df["Profit_per_Hectare_INR"] = (
    df["Profit_INR"] / df["Farm_Area_Hectares"]
)

# Production per hectare
df["Production_per_Hectare"] = (
    df["Production_Tonnes"] / df["Farm_Area_Hectares"]
)

print("New features created:")
print("- Profit_Margin_pct")
print("- Revenue_per_Hectare_INR")
print("- Cost_per_Hectare_INR")
print("- Profit_per_Hectare_INR")
print("- Production_per_Hectare")

display(df.head())


# ============================================================
# 14. OVERALL KPIs
# ============================================================

print("\n" + "=" * 70)
print("OVERALL AGRICULTURAL PERFORMANCE KPIs")
print("=" * 70)

kpis = {
    "Total Farms": df["Farm_ID"].nunique(),
    "Total Production (Tonnes)": df["Production_Tonnes"].sum(),
    "Average Yield (Tonnes/Ha)": df["Yield_Tonnes_Ha"].mean(),
    "Average Production (Tonnes)": df["Production_Tonnes"].mean(),
    "Average Revenue (INR)": df["Revenue_INR"].mean(),
    "Average Cost (INR)": df["Total_Cost_INR"].mean(),
    "Average Profit (INR)": df["Profit_INR"].mean(),
    "Average Water Usage (m³)": df["Water_Used_m3"].mean(),
    "Average Water Efficiency": df[
        "Water_Efficiency_t_per_1000m3"
    ].mean(),
    "Average Disease/Pest Risk (%)": df[
        "Disease_Pest_Risk_pct"
    ].mean()
}

kpi_df = pd.DataFrame(
    kpis.items(),
    columns=["KPI", "Value"]
)

display(kpi_df)


# ============================================================
# 15. SEASON DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("SEASON DISTRIBUTION")
print("=" * 70)

season_counts = df["Season"].value_counts()

display(season_counts)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=season_counts.index,
    y=season_counts.values
)

plt.title("Number of Farms by Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()


# ============================================================
# 16. SEASONAL PERFORMANCE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("SEASONAL AGRICULTURAL PERFORMANCE")
print("=" * 70)

season_summary = df.groupby("Season").agg(
    Farms=("Farm_ID", "count"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Production=("Production_Tonnes", "mean"),
    Avg_Revenue=("Revenue_INR", "mean"),
    Avg_Cost=("Total_Cost_INR", "mean"),
    Avg_Profit=("Profit_INR", "mean"),
    Avg_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    )
).round(2)

display(season_summary)


# ============================================================
# 17. AVERAGE YIELD BY SEASON
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean"
)

plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 18. YIELD DISTRIBUTION BY SEASON
# ============================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha"
)

plt.title("Yield Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 19. PRODUCTION BY SEASON
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Production_Tonnes",
    estimator="mean"
)

plt.title("Average Production by Season")
plt.xlabel("Season")
plt.ylabel("Average Production (Tonnes)")
plt.tight_layout()
plt.show()


# ============================================================
# 20. ECONOMIC PERFORMANCE BY SEASON
# ============================================================

print("\n" + "=" * 70)
print("ECONOMIC PERFORMANCE BY SEASON")
print("=" * 70)

economic_season = df.groupby("Season")[
    [
        "Revenue_INR",
        "Total_Cost_INR",
        "Profit_INR"
    ]
].mean().round(2)

display(economic_season)

economic_season.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Economic Performance by Season")
plt.xlabel("Season")
plt.ylabel("Amount (INR)")
plt.xticks(rotation=0)
plt.legend(
    title="Metric",
    labels=["Revenue", "Cost", "Profit"]
)

plt.tight_layout()
plt.show()


# ============================================================
# 21. PROFIT MARGIN BY SEASON
# ============================================================

profit_margin_season = df.groupby("Season")[
    "Profit_Margin_pct"
].mean()

print("\nAverage Profit Margin by Season:")
display(profit_margin_season.round(2))

plt.figure(figsize=(8, 5))

sns.barplot(
    x=profit_margin_season.index,
    y=profit_margin_season.values
)

plt.title("Average Profit Margin by Season")
plt.xlabel("Season")
plt.ylabel("Profit Margin (%)")
plt.tight_layout()
plt.show()


# ============================================================
# 22. ENVIRONMENTAL CONDITIONS
# ============================================================

print("\n" + "=" * 70)
print("ENVIRONMENTAL CONDITIONS ACROSS SEASONS")
print("=" * 70)

environment_summary = df.groupby("Season").agg(
    Rainfall=("Rainfall_mm", "mean"),
    Temperature=("Avg_Temperature_C", "mean"),
    Humidity=("Humidity_pct", "mean"),
    Sunlight=("Sunlight_Hours_Day", "mean"),
    Soil_Moisture=("Soil_Moisture_pct", "mean")
).round(2)

display(environment_summary)


# ============================================================
# 23. ENVIRONMENTAL CONDITIONS VISUALIZATION
# ============================================================

environment_summary.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Environmental Conditions Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend(title="Environmental Variable")
plt.tight_layout()
plt.show()


# ============================================================
# 24. RAINFALL VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Rainfall vs Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 25. TEMPERATURE VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Avg_Temperature_C",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Temperature vs Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 26. SOIL MOISTURE VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Soil_Moisture_pct",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Soil Moisture vs Yield")
plt.xlabel("Soil Moisture (%)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 27. CROP-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CROP-WISE AGRICULTURAL PERFORMANCE")
print("=" * 70)

crop_summary = df.groupby("Crop").agg(
    Farms=("Farm_ID", "count"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Production=("Production_Tonnes", "mean"),
    Avg_Revenue=("Revenue_INR", "mean"),
    Avg_Profit=("Profit_INR", "mean")
).sort_values(
    "Avg_Yield",
    ascending=False
).round(2)

display(crop_summary)


# ============================================================
# 28. AVERAGE YIELD BY CROP
# ============================================================

crop_order = (
    df.groupby("Crop")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="Crop",
    y="Yield_Tonnes_Ha",
    estimator="mean",
    order=crop_order
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 29. CROP × SEASON YIELD
# ============================================================

crop_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

print("\nAverage Yield by Crop and Season:")
display(crop_season_yield.round(2))

plt.figure(figsize=(10, 7))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 30. CROP × SEASON PRODUCTION
# ============================================================

crop_season_production = pd.pivot_table(
    df,
    values="Production_Tonnes",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    crop_season_production,
    annot=True,
    fmt=".2f"
)

plt.title("Average Production by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 31. IRRIGATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("IRRIGATION AND AGRICULTURAL PERFORMANCE")
print("=" * 70)

irrigation_summary = df.groupby(
    "Irrigation_Method"
).agg(
    Farms=("Farm_ID", "count"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Water_Used=("Water_Used_m3", "mean"),
    Avg_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    ),
    Avg_Profit=("Profit_INR", "mean")
).sort_values(
    "Avg_Yield",
    ascending=False
).round(2)

display(irrigation_summary)


# ============================================================
# 32. IRRIGATION VS YIELD
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Irrigation_Method",
    y="Yield_Tonnes_Ha"
)

plt.title("Yield Distribution by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# ============================================================
# 33. IRRIGATION VS WATER EFFICIENCY
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Irrigation_Method",
    y="Water_Efficiency_t_per_1000m3",
    estimator="mean"
)

plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# ============================================================
# 34. RESOURCE USAGE BY SEASON
# ============================================================

print("\n" + "=" * 70)
print("RESOURCE USAGE BY SEASON")
print("=" * 70)

resource_summary = df.groupby("Season").agg(
    Fertilizer=("Fertilizer_kg_ha", "mean"),
    Pesticide=("Pesticide_Litre_ha", "mean"),
    Nitrogen=("Nitrogen_kg_ha", "mean"),
    Phosphorus=("Phosphorus_kg_ha", "mean"),
    Potassium=("Potassium_kg_ha", "mean"),
    Water_Used=("Water_Used_m3", "mean")
).round(2)

display(resource_summary)


# ============================================================
# 35. WATER USAGE BY SEASON
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Water_Used_m3",
    estimator="mean"
)

plt.title("Average Water Usage by Season")
plt.xlabel("Season")
plt.ylabel("Water Used (m³)")
plt.tight_layout()
plt.show()


# ============================================================
# 36. WATER EFFICIENCY BY SEASON
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Water_Efficiency_t_per_1000m3",
    estimator="mean"
)

plt.title("Average Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency")
plt.tight_layout()
plt.show()


# ============================================================
# 37. FERTILIZER VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Fertilizer_kg_ha",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Fertilizer Usage vs Yield")
plt.xlabel("Fertilizer (kg/ha)")
plt.ylabel("Yield (Tonnes/ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 38. SEED QUALITY VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Seed_Quality_Score",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Seed Quality vs Yield")
plt.xlabel("Seed Quality Score")
plt.ylabel("Yield (Tonnes/ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 39. DISEASE AND PEST RISK
# ============================================================

print("\n" + "=" * 70)
print("DISEASE AND PEST RISK")
print("=" * 70)

risk_summary = df.groupby("Season").agg(
    Avg_Risk=("Disease_Pest_Risk_pct", "mean"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Profit=("Profit_INR", "mean")
).round(2)

display(risk_summary)


# ============================================================
# 40. DISEASE/PEST RISK BY SEASON
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Season",
    y="Disease_Pest_Risk_pct",
    estimator="mean"
)

plt.title("Average Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.tight_layout()
plt.show()


# ============================================================
# 41. DISEASE/PEST RISK VS YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Disease_Pest_Risk_pct",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Disease/Pest Risk vs Yield")
plt.xlabel("Disease/Pest Risk (%)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


# ============================================================
# 42. STATE-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("REGIONAL / STATE-WISE ANALYSIS")
print("=" * 70)

state_summary = df.groupby("State").agg(
    Farms=("Farm_ID", "count"),
    Avg_Yield=("Yield_Tonnes_Ha", "mean"),
    Avg_Production=("Production_Tonnes", "mean"),
    Avg_Revenue=("Revenue_INR", "mean"),
    Avg_Profit=("Profit_INR", "mean"),
    Avg_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    )
).sort_values(
    "Avg_Profit",
    ascending=False
).round(2)

display(state_summary)


# ============================================================
# 43. STATE-WISE YIELD
# ============================================================

state_order = (
    df.groupby("State")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="State",
    y="Yield_Tonnes_Ha",
    estimator="mean",
    order=state_order
)

plt.title("Average Yield by State")
plt.xlabel("State")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 44. STATE × SEASON YIELD
# ============================================================

state_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="State",
    columns="Season",
    aggfunc="mean"
)

print("\nAverage Yield by State and Season:")
display(state_season_yield.round(2))

plt.figure(figsize=(10, 7))

sns.heatmap(
    state_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by State and Season")
plt.xlabel("Season")
plt.ylabel("State")
plt.tight_layout()
plt.show()


# ============================================================
# 45. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

correlation_columns = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

correlation_matrix = df[
    correlation_columns
].corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Agricultural Variables")
plt.tight_layout()
plt.show()


# ============================================================
# 46. CORRELATION WITH YIELD
# ============================================================

print("\n" + "=" * 70)
print("VARIABLES CORRELATED WITH YIELD")
print("=" * 70)

yield_correlations = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .sort_values(ascending=False)
)

display(yield_correlations.round(3))


# ============================================================
# 47. CORRELATION WITH PROFIT
# ============================================================

print("\n" + "=" * 70)
print("VARIABLES CORRELATED WITH PROFIT")
print("=" * 70)

profit_correlations = (
    correlation_matrix["Profit_INR"]
    .sort_values(ascending=False)
)

display(profit_correlations.round(3))


# ============================================================
# 48. ANOVA TEST
# ============================================================

print("\n" + "=" * 70)
print("ONE-WAY ANOVA: YIELD ACROSS SEASONS")
print("=" * 70)

season_groups = [
    group["Yield_Tonnes_Ha"].values
    for _, group in df.groupby("Season")
]

anova_stat, anova_p = stats.f_oneway(*season_groups)

print("F-statistic:", round(anova_stat, 4))
print("p-value:", round(anova_p, 6))

if anova_p < 0.05:

    print(
        "\nConclusion:"
        "\nReject H0."
        "\nThere is a statistically significant difference "
        "in mean yield among the seasons."
    )

else:

    print(
        "\nConclusion:"
        "\nFail to reject H0."
        "\nThere is not enough statistical evidence to conclude "
        "that mean yield differs among the seasons."
    )


# ============================================================
# 49. PAIRWISE SEASONAL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("PAIRWISE SEASONAL COMPARISON")
print("=" * 70)

season_names = df["Season"].unique()

pairwise_results = []

for season1, season2 in combinations(season_names, 2):

    group1 = df[
        df["Season"] == season1
    ]["Yield_Tonnes_Ha"]

    group2 = df[
        df["Season"] == season2
    ]["Yield_Tonnes_Ha"]

    t_stat, p_value = stats.ttest_ind(
        group1,
        group2,
        equal_var=False
    )

    pairwise_results.append({
        "Season 1": season1,
        "Season 2": season2,
        "T-statistic": t_stat,
        "P-value": p_value,
        "Significant at 5%": "Yes" if p_value < 0.05 else "No"
    })

pairwise_df = pd.DataFrame(pairwise_results)

display(pairwise_df.round(4))


# ============================================================
# 50. CORRELATION SIGNIFICANCE TESTS
# ============================================================

print("\n" + "=" * 70)
print("CORRELATION SIGNIFICANCE TESTS")
print("=" * 70)

variables_of_interest = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_Moisture_pct",
    "Fertilizer_kg_ha",
    "Seed_Quality_Score",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]

correlation_results = []

for variable in variables_of_interest:

    r, p = stats.pearsonr(
        df[variable],
        df["Yield_Tonnes_Ha"]
    )

    correlation_results.append({
        "Variable": variable,
        "Correlation (r)": r,
        "P-value": p,
        "Significant at 5%": "Yes" if p < 0.05 else "No"
    })

correlation_results_df = pd.DataFrame(
    correlation_results
).sort_values(
    "Correlation (r)",
    ascending=False
)

display(correlation_results_df.round(4))


# ============================================================
# 51. PROFITABILITY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PROFITABILITY ANALYSIS")
print("=" * 70)

profitability_summary = df.groupby("Season").agg(
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Profit_Margin=("Profit_Margin_pct", "mean"),
    Average_Profit_per_Hectare=(
        "Profit_per_Hectare_INR",
        "mean"
    )
).round(2)

display(profitability_summary)


# ============================================================
# 52. PROFITABLE VS LOSS-MAKING FARMS
# ============================================================

df["Profit_Status"] = np.where(
    df["Profit_INR"] >= 0,
    "Profitable",
    "Loss-making"
)

profit_status = df["Profit_Status"].value_counts()

print("\nProfit Status:")
display(profit_status)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Profit_Status"
)

plt.title("Profitable vs Loss-Making Farms")
plt.xlabel("Profit Status")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()


# ============================================================
# 53. PROFITABILITY BY SEASON
# ============================================================

profit_status_season = pd.crosstab(
    df["Season"],
    df["Profit_Status"],
    normalize="index"
) * 100

print("\nProfitability Status by Season (%):")
display(profit_status_season.round(2))

profit_status_season.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 6)
)

plt.title("Profitability Status by Season")
plt.xlabel("Season")
plt.ylabel("Percentage of Farms")
plt.xticks(rotation=0)
plt.legend(title="Profit Status")
plt.tight_layout()
plt.show()


# ============================================================
# 54. OUTLIER DETECTION
# ============================================================

print("\n" + "=" * 70)
print("OUTLIER DETECTION")
print("=" * 70)

outlier_columns = [
    "Rainfall_mm",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3"
]

outlier_summary = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %": len(outliers) / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df.round(2))

print(
    "\nNote: Outliers are reported rather than automatically removed "
    "because extreme agricultural observations may represent genuine "
    "real-world conditions."
)


# ============================================================
# 55. YIELD DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Yield_Tonnes_Ha",
    bins=50,
    kde=True
)

plt.title("Distribution of Agricultural Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# ============================================================
# 56. LOG-TRANSFORMED DISTRIBUTIONS
# ============================================================

skewed_columns = [
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for col in skewed_columns:

    plt.figure(figsize=(9, 5))

    sns.histplot(
        np.log1p(df[col]),
        bins=40,
        kde=True
    )

    plt.title(
        f"Log-Transformed Distribution of {col}"
    )

    plt.xlabel(
        f"log(1 + {col})"
    )

    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# 57. TOP PERFORMING CROPS
# ============================================================

print("\n" + "=" * 70)
print("TOP PERFORMING CROPS")
print("=" * 70)

top_crops = (
    df.groupby("Crop")
    .agg(
        Average_Yield=("Yield_Tonnes_Ha", "mean"),
        Average_Profit=("Profit_INR", "mean"),
        Average_Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        )
    )
    .sort_values(
        "Average_Profit",
        ascending=False
    )
)

display(top_crops.round(2))


# ============================================================
# 58. BEST CROP-SEASON COMBINATIONS
# ============================================================

print("\n" + "=" * 70)
print("BEST CROP-SEASON COMBINATIONS")
print("=" * 70)

crop_season_summary = (
    df.groupby(["Crop", "Season"])
    .agg(
        Farms=("Farm_ID", "count"),
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Profit=("Profit_INR", "mean"),
        Avg_Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        )
    )
    .sort_values(
        "Avg_Profit",
        ascending=False
    )
)

display(
    crop_season_summary.round(2).head(15)
)


# ============================================================
# 59. BEST STATE-SEASON COMBINATIONS
# ============================================================

print("\n" + "=" * 70)
print("BEST STATE-SEASON COMBINATIONS")
print("=" * 70)

state_season_summary = (
    df.groupby(["State", "Season"])
    .agg(
        Farms=("Farm_ID", "count"),
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Profit=("Profit_INR", "mean"),
        Avg_Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        )
    )
    .sort_values(
        "Avg_Profit",
        ascending=False
    )
)

display(
    state_season_summary.round(2).head(15)
)


# ============================================================
# 60. SEASONAL RANKING
# ============================================================

print("\n" + "=" * 70)
print("SEASONAL PERFORMANCE RANKING")
print("=" * 70)

season_ranking = (
    df.groupby("Season")
    .agg(
        Yield=("Yield_Tonnes_Ha", "mean"),
        Profit=("Profit_INR", "mean"),
        Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        ),
        Profit_Margin=("Profit_Margin_pct", "mean"),
        Disease_Risk=("Disease_Pest_Risk_pct", "mean")
    )
)

season_ranking["Yield_Rank"] = (
    season_ranking["Yield"]
    .rank(ascending=False)
)

season_ranking["Profit_Rank"] = (
    season_ranking["Profit"]
    .rank(ascending=False)
)

season_ranking["Water_Efficiency_Rank"] = (
    season_ranking["Water_Efficiency"]
    .rank(ascending=False)
)

season_ranking["Profit_Margin_Rank"] = (
    season_ranking["Profit_Margin"]
    .rank(ascending=False)
)

season_ranking["Disease_Risk_Rank"] = (
    season_ranking["Disease_Risk"]
    .rank(ascending=True)
)

display(season_ranking.round(2))


# ============================================================
# 61. AUTOMATIC KEY FINDINGS
# ============================================================

print("\n" + "=" * 70)
print("AUTOMATIC KEY FINDINGS")
print("=" * 70)

best_yield_season = season_summary["Avg_Yield"].idxmax()
best_profit_season = season_summary["Avg_Profit"].idxmax()
best_water_season = season_summary[
    "Avg_Water_Efficiency"
].idxmax()

lowest_risk_season = risk_summary[
    "Avg_Risk"
].idxmin()

best_yield_value = season_summary.loc[
    best_yield_season,
    "Avg_Yield"
]

best_profit_value = season_summary.loc[
    best_profit_season,
    "Avg_Profit"
]

best_water_value = season_summary.loc[
    best_water_season,
    "Avg_Water_Efficiency"
]

lowest_risk_value = risk_summary.loc[
    lowest_risk_season,
    "Avg_Risk"
]

print(
    f"\n1. Highest average yield: "
    f"{best_yield_season} "
    f"({best_yield_value:.2f} Tonnes/Ha)"
)

print(
    f"2. Highest average profit: "
    f"{best_profit_season} "
    f"(INR {best_profit_value:,.2f})"
)

print(
    f"3. Highest water efficiency: "
    f"{best_water_season} "
    f"({best_water_value:.2f})"
)

print(
    f"4. Lowest disease/pest risk: "
    f"{lowest_risk_season} "
    f"({lowest_risk_value:.2f}%)"
)


# ============================================================
# 62. BEST CROP BY YIELD
# ============================================================

best_crop = crop_summary["Avg_Yield"].idxmax()
best_crop_yield = crop_summary.loc[
    best_crop,
    "Avg_Yield"
]

print(
    f"\n5. Highest average-yield crop: "
    f"{best_crop} "
    f"({best_crop_yield:.2f} Tonnes/Ha)"
)


# ============================================================
# 63. BEST IRRIGATION METHOD
# ============================================================

best_irrigation = irrigation_summary[
    "Avg_Water_Efficiency"
].idxmax()

best_irrigation_efficiency = irrigation_summary.loc[
    best_irrigation,
    "Avg_Water_Efficiency"
]

print(
    f"6. Highest water-efficiency irrigation method: "
    f"{best_irrigation} "
    f"({best_irrigation_efficiency:.2f})"
)


# ============================================================
# 64. STRONGEST POSITIVE / NEGATIVE YIELD CORRELATIONS
# ============================================================

yield_corr_without_self = yield_correlations.drop(
    "Yield_Tonnes_Ha"
)

strongest_positive = yield_corr_without_self.idxmax()
strongest_negative = yield_corr_without_self.idxmin()

print(
    f"\n7. Strongest positive correlation with yield: "
    f"{strongest_positive} "
    f"({yield_corr_without_self[strongest_positive]:.3f})"
)

print(
    f"8. Strongest negative correlation with yield: "
    f"{strongest_negative} "
    f"({yield_corr_without_self[strongest_negative]:.3f})"
)


# ============================================================
# 65. FINAL SUMMARY TABLE
# ============================================================

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

final_summary = pd.DataFrame({
    "Metric": [
        "Highest Yield Season",
        "Highest Profit Season",
        "Highest Water Efficiency Season",
        "Lowest Disease/Pest Risk Season",
        "Highest Yield Crop",
        "Most Water-Efficient Irrigation Method",
        "ANOVA Significant?"
    ],
    "Result": [
        best_yield_season,
        best_profit_season,
        best_water_season,
        lowest_risk_season,
        best_crop,
        best_irrigation,
        "Yes" if anova_p < 0.05 else "No"
    ]
})

display(final_summary)


# ============================================================
# 66. SAVE CLEANED DATASET
# ============================================================

cleaned_file = (
    "/content/cleaned_seasonal_agriculture_dataset.csv"
)

df.to_csv(
    cleaned_file,
    index=False
)

print(
    "\nCleaned dataset saved as:"
    "\ncleaned_seasonal_agriculture_dataset.csv"
)


# ============================================================
# 67. DOWNLOAD CLEANED DATASET
# ============================================================

print("\nYou can download the cleaned dataset below.")

files.download(
    "cleaned_seasonal_agriculture_dataset.csv"
)


# ============================================================
# END OF PROJECT
# ============================================================

print("\n" + "=" * 70)
print("SEASONAL AGRICULTURE PERFORMANCE ANALYSIS COMPLETED")
print("=" * 70)
```
